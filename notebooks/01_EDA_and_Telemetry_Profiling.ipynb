# 01. Exploratory Data Analysis (EDA) - AI4I 2020 Predictive Maintenance

**Dataset**: AI4I 2020 Predictive Maintenance Dataset (10,000 records, 14 attributes).

### Objectives:
1. Sensor telemetry distributions and baseline statistical profiling.
2. Multimodal failure rate distribution and class imbalance inspection.
3. Physical feature engineering (Thermal Delta, Mechanical Power, Overstrain Product).
4. Pearson & Spearman correlation heatmaps.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='darkgrid')
df = pd.read_csv('../data/ai4i2020.csv')
print(f'Dataset Shape: {df.shape}')
df.head()
```

```python
print('--- Dataset Summary ---')
print(df.info())
print('\n--- Missing Values ---')
print(df.isnull().sum())
print('\n--- Target Distribution ---')
print(df['Machine failure'].value_counts(normalize=True))
```

```python
# Physical Feature Engineering
df['Temp_diff'] = df['Process temperature [K]'] - df['Air temperature [K]']
df['Power_kW'] = (df['Torque [Nm]'] * df['Rotational speed [rpm]'] * 2 * np.pi) / 60000
df['Overstrain'] = df['Tool wear [min]'] * df['Torque [Nm]']

print('Engineered Features Created Successfully:')
df[['Temp_diff', 'Power_kW', 'Overstrain']].describe()
```

```python
plt.figure(figsize=(10, 6))
sns.heatmap(df.select_dtypes(include=[np.number]).corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Sensor Telemetry Correlation Matrix')
plt.tight_layout()
plt.show()
```
