```python
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestClassifier
import joblib

df = pd.read_csv('../data/ai4i2020.csv')

def get_failure_mode(row):
    if row['Machine failure'] == 0:
        return 'No Failure'
    if row['TWF'] == 1:
        return 'Tool Wear Failure'
    if row['HDF'] == 1:
        return 'Heat Dissipation Failure'
    if row['PWF'] == 1:
        return 'Power Failure'
    if row['OSF'] == 1:
        return 'Overstrain Failure'
    if row['RNF'] == 1:
        return 'Random Failure'
    return 'Other Failure'

df['Failure_Type'] = df.apply(get_failure_mode, axis=1)
print('Failure Type Breakdown:')
print(df['Failure_Type'].value_counts())
```
